```python
import pandas as pd

df = pd.DataFrame({
    'date_strings': ['2023-09-15', '2023/10/26', '15.11.2023']
})

# Pandas will attempt to infer the format for each entry
df['datetime_column'] = pd.to_datetime(df['date_strings'], format='mixed')
print(df['datetime_column'] < pd.Timestamp.now())  # Check for future dates
```
